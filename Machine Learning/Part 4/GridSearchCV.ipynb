```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn5 = KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn5.fit(X_train, y_train)
```

```python
model_knn5.score(X_test, y_test)
```

```python
# Changing the value of n_neighbors=9
model_knn9 = KNeighborsClassifier(n_neighbors=9)
```

```python
model_knn9.fit(X_train, y_train)
```

```python
model_knn9.score(X_test, y_test)
```

```python
# Changing the value of n_neighbors=13
model_knn13 = KNeighborsClassifier(n_neighbors=13)
```

```python
model_knn13.fit(X_train, y_train)
```

```python
model_knn13.score(X_test, y_test)
```

# SVM

```python
from sklearn.svm import SVC
```

```python
model_svm_nopara = SVC(gamma='auto')
```

```python
model_svm_nopara.fit(X_train, y_train)
```

```python
model_svm_nopara.score(X_test, y_test)
```

```python
# Using parameters
model_svm = SVC(C=30, kernel='rbf', gamma='auto')
```

```python
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)
```

```python
# Changing the value of kernel='linear'
model_svm_kernel = SVC(C=30, kernel='linear', gamma='auto')
```

```python
model_svm_kernel.fit(X_train, y_train)
```

```python
model_svm_kernel.score(X_test, y_test)
```

```python
# Changing the value of C=10
model_svm_C = SVC(C=10, kernel='linear', gamma='auto')
```

```python
model_svm_C.fit(X_train, y_train)
```

```python
model_svm_C.score(X_test, y_test)
```

# Grid Search CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm_nopara), {
    'C': [1, 10, 20, 30],
    'kernel': ['rbf', 'linear'],
}, cv=5, return_train_score=False)
```

```python
classifier.fit(X, y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C', 'param_kernel', 'mean_test_score']]
```

# Randam Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r = RandomizedSearchCV((model_svm_nopara), {
    'C': [1, 10, 20, 30],
    'kernel': ['rbf', 'linear'],
}, n_iter=4, cv=5, return_train_score=False)
```

```python
classifier_r.fit(X, y)
```

```python
results_r = pd.DataFrame(classifier_r.cv_results_)
```

```python
results_r[['param_C', 'param_kernel', 'mean_test_score']]
```
