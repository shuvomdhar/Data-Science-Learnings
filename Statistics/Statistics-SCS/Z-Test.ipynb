# Z-TEST

```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))
z_score
```

# P-TEST

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("I will reject the null hypothesis")
else:
    print("I will accept the null hypothesis")
```

```python
samp = [0.62, 0.65, 0.67, 0.64, 0.66, 0.68, 0.69, 0.70, 0.71, 0.73, 0.72, 0.74, 0.75, 0.76, 0.77, 0.78, 0.79, 0.80, 0.81, 0.82, 0.83, 0.84, 0.85, 0.86, 0.87, 0.88, 0.89, 0.90, 0.91, 0.92]
popul_mean = 0.6
popul_std = 0.1
samp_mean = np.mean(samp)
n = len(samp)
```

```python
zscore = (samp_mean - popul_mean) / (popul_std / np.sqrt(n))
zscore
```

```python
pvalue = 2 * (1 - norm.cdf(abs(zscore)))
pvalue
```
