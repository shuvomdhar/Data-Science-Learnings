```python
import numpy as np
import seaborn as sns
from scipy.stats import f_oneway
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
```

```python
df
```

```python
df = df[['age', 'pclass']].dropna()
```

```python
df['pclass'].unique()
```

```python
class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
```

```python
print("Class 1")
class_1
```

```python
print("Class 2")
class_2
```

```python
print("Class 3")
class_3
```

```python
f_stats, p_value = f_oneway(class_1, class_2, class_3)
```

```python
f_stats
```

```python
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("We will reject the null hypothesis and there is a significant difference in atleast one passenger class")
else:
    print("There is no connection between sex and servived")
```
