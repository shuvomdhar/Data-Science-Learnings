```python
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
```

```python
df
```

```python
contingency_table = pd.crosstab(df['sex'], df['survived'])
```

```python
contingency_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)
```

```python
expected
```

```python
dof
```

```python
p_value
```

```python
chi2
```

```python
alpha = 0.05

if p_value < alpha:
    print("We will reject the null hypothesis and there is a significant relationship between sex and servived")
else:
    print("There is no connection between sex and servived")
```
