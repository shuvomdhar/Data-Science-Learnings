```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x, y)

plt.show()
```

```python
x = ['Mon', 'Tues', 'Wed', 'Thurs', 'Fri'] # x-axis
y = [10, 15, 7, 20, 12] # y-axis

plt.plot(x, y)

plt.title('Bakery Sales This Week')

plt.xlabel('Day of the Week')
plt.ylabel('Sales per Day')

plt.show()
```

```python
months = [1,2,3,4]
sales = [1000,1500,1200,1800]

plt.plot(months, sales, color='blue', linestyle='--', linewidth=2, marker='o', label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per Month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray', linestyle=':', linewidth=1)
plt.xlim(1, 4)
plt.ylim(0, 2000)
plt.xticks([1,2,3,4], ['M1', 'M2', 'M3', 'M4'])
plt.show()
```

```python
product = ['A', 'B', 'C', 'D']
sales = [1000, 1500, 800, 1200]

plt.bar(product, sales, color='orange', label='Sales 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sale Comparison')
plt.legend()
plt.show()

plt.barh(product, sales, color='skyblue', label='Sales 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sale Comparison')
plt.legend()
plt.show()
```

```python
regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]

plt.pie(revenue, labels=regions, autopct='%1.1f%%', colors=['gold', 'skyblue', 'lightgreen', 'coral'])
plt.title('Revenue Contribution by Region')
plt.show()
```

```python
scores = [45, 67, 89, 56, 78, 88, 92, 60, 74, 81, 55, 66, 75, 82, 90, 85, 70, 73, 68, 77]

plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.xlabel('Score Range')
plt.ylabel('Number of students')
plt.title('Score Distribution of Students')
plt.show()
```

```python
hours_studies = [1,2,3,4,5,6,7,8]
exam_scores = [50, 55, 60, 65, 70, 75, 80, 85]

# scatter plot
plt.scatter(hours_studies, exam_scores, color='red', marker='^', label='Student Data')

plt.xlabel('Hours studied')
plt.ylabel('Exam Score')
plt.title('Relationship between Study Time and Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# scatter plot
plt.scatter([1,2,3], [50,55,60], color='blue', label='Class A') #g1
plt.scatter([1,2,3], [45,50,52], color='orange', label='Class B') #g2

plt.xlabel('Hours studied')
plt.ylabel('Exam Score')
plt.title('Relationship between Study Time and Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(1, 2, 1) # 1-row , 2-column , 1st subplot
plt.plot(x, y)
plt.title("Line Chart")

plt.subplot(1, 2, 2) # 1-row , 2-column , 2nd subplot
plt.bar(x,y)
plt.title('Bar Chart')

plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x, y, color='blue')
ax[0].set_title('Line Plot')

ax[1].bar(x, y, color='green')
ax[1].set_title('Bar Chart')

fig.suptitle('Comparison of Line Plot and Bar Chart')

plt.tight_layout()
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, color='blue', marker='o')
plt.title('Simple Line Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.savefig('line_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
